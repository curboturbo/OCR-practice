```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class CNN(nn.Module):
  def __init__(self):
    super(CNN, self).__init__()
    self.conv1 = nn.Conv2d(in_channels=1, out_channels=32,kernel_size=3,padding=1)
    self.conv2 = nn.Conv2d(32,64,kernel_size=3,padding=1)
    self.pool = nn.MaxPool2d(2,2)
    self.fc1 = nn.Linear(64*14*14, 128)
    self.fc2 = nn.Linear(128, 10)

  def forward(self, x):
    x = F.relu(self.conv1(x))
    x = F.relu(self.conv2(x))
    x = self.pool(x)
    x = x.view(-1,64*14*14)
    x = F.relu(self.fc1(x))
    return self.fc2(x)
```

```python
model = CNN()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
criterion = nn.CrossEntropyLoss()
```

```python
from torchvision import datasets, transforms
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081))])
```

```python
from torchvision import datasets, transforms
train_dataset = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root="./data", train=False, download=True, transform=transform)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
test_dataset
```

```python
train_dataloader = DataLoader(train_dataset, batch_size=64, shuffle=True)
```

```python
test_dataloader = DataLoader(test_dataset, batch_size=1000,shuffle=False)
```

```python
def train(model, loader, optimizer, criterion, epoch):
  model.train()
  for batch_idx, (data,target) in enumerate(loader):
    optimizer.zero_grad()
    output = model(data)
    loss = criterion(output,target)
    loss.backward()
    optimizer.step()
```

```python
for epoch in range(1,6):
  train(model, train_dataloader, optimizer, criterion, epoch)
```

```python
def test(model, loader):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for data, target in loader:
            output = model(data)
            pred = output.argmax(dim=1)
            correct += (pred == target).sum().item()
            total += target.size(0)

    acc = 100. * correct / total
    print(f"Test accuracy: {acc:.2f}%")
```

```python
test(model, test_dataloader)
```

```python
from ultralytics import YOLO
import torch
import cv2
import easyocr
import numpy as np


path_model = './best.pt'
path_photo = './bio.png'


model = YOLO(path_model)

class_names = model.names

img = cv2.imread(path_photo)
if img is None:
    print(f"Ошибка: Не удалось загрузить изображение по пути {path_photo}")
    exit()


results_yolo = model(path_photo, save=False, conf=0.2)

extracted_texts = []


reader = easyocr.Reader(['ru'])

for res in results_yolo:
    for *xyxy, conf, cls_idx in res.boxes.data.tolist():
        x1, y1, x2, y2 = map(int, xyxy)
        cls_idx = int(cls_idx)
        class_name = class_names[cls_idx] if cls_idx < len(class_names) else f"Unknown_Class_{cls_idx}"
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(img.shape[1], x2)
        y2 = min(img.shape[0], y2)
        if x2 > x1 and y2 > y1:
            cropped_text_region = img[y1:y2, x1:x2]

            if cropped_text_region.size == 0:
                print(f"Предупреждение: Пустая область обрезки для рамки ({x1}, {y1}, {x2}, {y2}). Пропускаем.")
                continue

            try:
                text_results = reader.readtext(cropped_text_region)

                full_text = ""
                for (bbox_easyocr, text, prob) in text_results:
                    full_text += text + " "

                cleaned_text = full_text.strip()

                if cleaned_text:
                    extracted_texts.append({
                        'bbox': (x1, y1, x2, y2),
                        'class_name': class_name, # Добавляем название класса
                        'text': cleaned_text
                    })
                    # Выводим название класса, а затем распознанный текст
                    print(f"Обнаружено: {class_name} в рамке ({x1}, {y1}, {x2}, {y2}):\n{cleaned_text}\n---")

            except Exception as e:
                print(f"Ошибка при обработке ROI ({x1}, {y1}, {x2}, {y2}): {e}")
                continue
        else:
            print(f"Предупреждение: Невалидная рамка после коррекции ({x1}, {y1}, {x2}, {y2}). Пропускаем.")


print("\n--- Все извлеченные тексты ---")
if not extracted_texts:
    print("Тексты не найдены.")
else:
    for item in extracted_texts:
        # Выводим название класса, а затем текст
        print(f"Bounding Box: {item['bbox']}, Class: {item['class_name']}, Text: {item['text']}")
```

```python
from ultralytics import YOLO
import cv2
import pytesseract
import numpy as np
from PIL import Image
import os

path_model = './best.pt'

path_photo = './bio.png'



try:
    model = YOLO(path_model)
    class_names = model.names
    print(f"Модель YOLO '{path_model}' успешно загружена. Обнаруженные классы: {class_names}")
except Exception as e:
    print(f"Ошибка при загрузке модели YOLO: {e}")
    exit()

img = cv2.imread(path_photo)
if img is None:
    print(f"Ошибка: Не удалось загрузить изображение по пути {path_photo}. Проверьте путь и целостность файла.")
    exit()
print(f"Изображение '{path_photo}' успешно загружено. Размеры: {img.shape[1]}x{img.shape[0]}")


print("Запуск детекции объектов YOLO...")
results_yolo = model(path_photo, save=False, conf=0.2)
print("Детекция объектов завершена.")

extracted_texts = []

for res in results_yolo:
    for *xyxy, conf, cls_idx in res.boxes.data.tolist():
        x1, y1, x2, y2 = map(int, xyxy)
        cls_idx = int(cls_idx)

        class_name = class_names[cls_idx] if cls_idx < len(class_names) else f"Unknown_Class_{cls_idx}"

        print(f"\n--- Обнаружен объект: {class_name} с уверенностью {conf:.2f} в рамке ({x1}, {y1}, {x2}, {y2}) ---")
        x1 = max(0, x1)
        y1 = max(0, y1)
        x2 = min(img.shape[1], x2)
        y2 = min(img.shape[0], y2)

        if x2 <= x1 or y2 <= y1:
            print(f"Предупреждение: Невалидная или нулевая область для обрезки после коррекции ({x1}, {y1}, {x2}, {y2}). Пропускаем.")
            continue

        cropped_text_region = img[y1:y2, x1:x2]

        if cropped_text_region.size == 0:
            print(f"Предупреждение: Обрезанная область оказалась пустой для рамки ({x1}, {y1}, {x2}, {y2}). Пропускаем.")
            continue

        try:
            cropped_rgb = cv2.cvtColor(cropped_text_region, cv2.COLOR_BGR2RGB)

            pil_image = Image.fromarray(cropped_rgb)

            text_results = pytesseract.image_to_string(pil_image, lang='rus')

            cleaned_text = text_results.strip()

            if cleaned_text:
                extracted_texts.append({
                    'bbox': (x1, y1, x2, y2),
                    'class_name': class_name,
                    'text': cleaned_text
                })
                print(f"Распознанный текст:\n{cleaned_text}")
            else:
                print("Текст не распознан в этой области.")

        except Exception as e:
            print(f"Ошибка при распознавании текста Tesseract для ROI ({x1}, {y1}, {x2}, {y2}): {e}")
            continue
print("\n--- Сводка: Все извлеченные тексты ---")
if not extracted_texts:
    print("Тексты не найдены ни в одной из обнаруженных областей.")
else:
    for item in extracted_texts:
        print(f"Класс: {item['class_name']}")
        print(f"Рамка (x1, y1, x2, y2): {item['bbox']}")
        print(f"Текст: {item['text']}")
        print("-" * 30)
```

```python
import pandas as pd
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments
from datasets import Dataset
from sklearn.model_selection import train_test_split
import torch
"""ПРИМЕР НЕПРАВИЛЬНОГО ОБУЧЕНИЯ BERT БЕЗ ТОКЕНИЗАТОРА НА РУССКОМ"""
file_path = './bert.xlsx'
df = pd.read_excel(file_path, header=None) 
df.columns = ['Задание', 'Ответ', 'Оценка']  

df['Оценка'] = df['Оценка'].astype(int)


df['text'] = "Задание №" + df['Задание'].astype(str) + " " + df['Ответ'].astype(str)


train_data, test_data = train_test_split(df, test_size=0.2, random_state=42)


train_dataset = Dataset.from_pandas(train_data.reset_index(drop=True))
test_dataset = Dataset.from_pandas(test_data.reset_index(drop=True))
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

def tokenize_function(examples):
    return tokenizer(examples['text'], padding="max_length", truncation=True, max_length=128)

train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)


train_dataset = train_dataset.rename_column("Оценка", "labels")
test_dataset = test_dataset.rename_column("Оценка", "labels")


cols_to_remove = ["Задание", "Ответ", "text"]
train_dataset = train_dataset.remove_columns(cols_to_remove)
test_dataset = test_dataset.remove_columns(cols_to_remove)


train_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'labels'])
test_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'labels'])


labels = df['Оценка'].unique()
if min(labels) != 0:
    df['Оценка'] = df['Оценка'] - min(labels)

num_labels = df['Оценка'].nunique()
print(f"Обнаружено уникальных меток: {df['Оценка'].unique()}. Количество меток (num_labels): {num_labels}")

model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=num_labels)


training_args = TrainingArguments(
    output_dir='./results',          
    num_train_epochs=3,              
    per_device_train_batch_size=8,   
    per_device_eval_batch_size=8,    
    warmup_steps=500,                
    weight_decay=0.01,               
    logging_dir='./logs',            
    logging_steps=10,
    report_to="none",                
    save_steps=500,                  
    eval_steps=500,                  
)


trainer = Trainer(
    model=model,                         
    args=training_args,                  
    train_dataset=train_dataset,         
    eval_dataset=test_dataset,           
)


trainer.train()


results = trainer.evaluate()
print(results)
```
